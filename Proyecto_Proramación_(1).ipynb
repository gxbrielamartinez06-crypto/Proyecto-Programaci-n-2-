# Dinero: `Valor`, `Salario` y `Clausula`
Como lo recalcamos al inicio algunas columnas como `Valor` guardan textos como `"€67.5M"` o `"€560K"`. Si bien, entendemos que es dinero, pandas los ve como texto.
* *Consecuencias:* No poder calcular promedios, sumas ni boxplots con ellos. Para poder detectar datos faltantes y outliers, primero hay que convertirlos a números.

* *Solución:* Por medio de una función, miraremos la última letra: si es `M` multiplica por 1.000.000, si es `K` por 1.000.

**Observación:** Se identificó que *€0*, no es un valor real (ningún jugador profesional vale o gana exactamente 0 €). Significa que **"no hay dato"**.

```python
def convertir_dinero(texto):
    ultimo = texto[-1]
    if ultimo == "M":
        numero = float(texto[1:-1]) * 1000000
    elif ultimo == "K":
        numero = float(texto[1:-1]) * 1000
    else:
        numero = float(texto[1:])
    if numero == 0:
        return None
    return numero
```

```python
df_nuevo["Valor"] = df_nuevo["Valor"].map(convertir_dinero)
df_nuevo["Salario"] = df_nuevo["Salario"].map(convertir_dinero)
df_nuevo["Clausula"] = df_nuevo["Clausula"].map(convertir_dinero)

df_nuevo[["Valor", "Salario", "Clausula"]].head(5)
```

# Medidas corporales: `Estatura` y `Peso`

Ahora trabajaremos en la estatura y peso:
* *Medidas de estatura:*  Pulgadas y pies - Ejemplo: (`5'7"`)  
* *Peso:* Libras - Ejemplo: (`159lbs`).
* *Medidas empleadas en Colombia:* **Centímetros** y **kilogramos**.
* *Decisión:* Crear columnas nuevas (`Estatura_cm`, `Peso_kg`) y eliminamos las viejas con `drop(columns=...)`

```python
def convertir_estatura(texto):
    pies = int(texto[0])
    pulgadas = int(texto[2:-1])
    return round((pies * 12 + pulgadas) * 2.54,2)

def convertir_peso(texto):
    libras = int(texto[:-3])
    return round(libras * 0.4536,2)
```

```python
df_nuevo["Estatura_cm"] = df_nuevo["Estatura"].map(convertir_estatura)
df_nuevo["Peso_kg"] = df_nuevo["Peso"].map(convertir_peso)
df_nuevo = df_nuevo.drop(columns=["Estatura", "Peso"])
df_nuevo[["Estatura_cm", "Peso_kg"]].head(6)
```

# Legibilidad: `Posición`, `Pais`, `Pie`, `Ritmos (Ataque/Defensa)`

* Para `Ritmo_ataque`,`Ritmo_defensa`y `Pie`, se empleo una función y posteriormente un .map()

* Para `Posicion`, se ha usado un diccionario.

* Pero, para `Pais`, por la magnitud de información (más de 90 paises), fue muy complicado crear las mismas opciones que con las categorias anteriores, por ellos, con uso de IA se creo uno, que hiciera esto mismo.

```python
from babel import Locale

en = Locale('en').territories
es = Locale('es').territories
mapa = {en[k]: es[k] for k in en if k in es}

mapa.update({
    'England': 'Inglaterra', 'Scotland': 'Escocia', 'Wales': 'Gales',
    'Northern Ireland': 'Irlanda del Norte', 'Republic of Ireland': 'Irlanda',
    'Korea Republic': 'Corea del Sur', 'Korea DPR': 'Corea del Norte',
    'China PR': 'China', 'United States': 'Estados Unidos',
    'Ivory Coast': 'Costa de Marfil', 'DR Congo': 'RD del Congo',
    'Cape Verde': 'Cabo Verde', 'Turkey': 'Turquía',
    'Bosnia Herzegovina': 'Bosnia y Herzegovina',
    'Central African Republic': 'República Centroafricana',
    'Trinidad & Tobago': 'Trinidad y Tobago',
    'Antigua & Barbuda': 'Antigua y Barbuda',
    'Guinea Bissau': 'Guinea-Bisáu'
})

df_nuevo.columns = df_nuevo.columns.str.strip()
col = next(c for c in df_nuevo.columns if c.lower() in ['pais', 'país', 'nationality'])
df_nuevo[col] = df_nuevo[col].replace(mapa)
```

```python
def ritmo(clasificacion):
    if clasificacion=="High":
        return "Alto"

    elif clasificacion=="Medium":
        return "Medio"

    else:
        return "Bajo"

def foot(pie):
    if pie=="Left":
        return "Izquierdo"

    else:
        return "Derecho"
```

```python
df_nuevo["Pie_dominante"] = df_nuevo["Pie_dominante"].map(foot)
df_nuevo["Ritmo_defensa"] = df_nuevo["Ritmo_defensa"].map(ritmo)
df_nuevo["Ritmo_ataque"] = df_nuevo["Ritmo_ataque"].map(ritmo)
```

```python
posiciones = {
    'GK': 'Portero',
    'CB': 'Defensa central',
    'LB': 'Lateral izquierdo',
    'RB': 'Lateral derecho',
    'LWB': 'Carrilero izquierdo',
    'RWB': 'Carrilero derecho',
    'CDM': 'Mediocentro defensivo',
    'CM': 'Mediocentro',
    'CAM': 'Mediocentro ofensivo',
    'LM': 'Volante izquierdo',
    'RM': 'Volante derecho',
    'LW': 'Extremo izquierdo',
    'RW': 'Extremo derecho',
    'CF': 'Mediapunta',
    'ST': 'Delantero centro'
}

df_nuevo['Posicion'] = df_nuevo['Posicion'].map(posiciones)
```

## Equipo
* Los datos de la varible equipo estan llenas de "\n" (o espacios) y fechas no importantes para este análisis, esto nos lleva a dejar unicamente el nombre del equipo del jugador.

```python
def equiporeal(contrato):
    texto2=contrato.strip()
    equipo=""

    for letra in texto2:
        if letra == "\n":
            break
        equipo +=letra

    return equipo
```

```python
df_nuevo["Equipo"]=df_nuevo["Equipo"].map(equiporeal)
```

>**Maximo de edad en porteros**

>**Los jugadores con un ritmo alto de ataque y de defensa**

```python
completos = df_nuevo[(df_nuevo["Ritmo_ataque"] == "Alto") & (df_nuevo["Ritmo_defensa"] == "Alto")]

porcent_completos = len(completos) / len(df_nuevo) * 100
print(f"Jugadores completos: {len(completos)} de {len(df_nuevo)} ({porcent_completos:.1f} %)")
print(f"Overall medio de los completos: {completos['Overall'].mean():.1f}")
print(f"Overall medio de toda la base: {df_nuevo['Overall'].mean():.1f}")
print(f"Valor mediano de los completos: {completos['Valor'].median() / 1000000:.2f} millones")
print(f"Valor mediano de toda la base: {df_nuevo['Valor'].median() / 1000000:.2f} millones")

medio = ["Mediocentro", "Mediocentro defensivo", "Mediocentro ofensivo", "Volante derecho", "Volante izquierdo"]
mediocampistas = completos["Posicion"].isin(medio).sum()
porcent_medio = mediocampistas / len(completos) * 100
print(f"Mediocampistas: {mediocampistas} de {len(completos)} ({porcent_medio:.1f} %)")
```

Los jugadores completos tienen un nivel mayor que el resto: su overall promedio es 69,9 frente a 65,7, y el 51 % de ellos tiene un overall de 70 o más, frente al 28 % en toda la base.
